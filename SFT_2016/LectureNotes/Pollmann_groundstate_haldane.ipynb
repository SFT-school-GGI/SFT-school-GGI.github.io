```python
import numpy as np
import pylab as pl
from scipy.linalg import expm as expm
from scipy.sparse.linalg.eigen.arpack import eigs
from numpy.linalg import svd
```

```python
def update(G,l,U,chi):
    " Updates the G and l matrices using U_bond and the TEBD protocol "
    d = U.shape[0]
    for step in [0,1]:
        A = np.mod(step,2); B = np.mod(step+1,2)

        # Construct theta        
        theta = np.tensordot(np.diag(l[B]),G[A],axes=(1,1))
        theta = np.tensordot(theta,np.diag(l[A],0),axes=(2,0))
        theta = np.tensordot(theta,G[B],axes=(2,1))
        theta = np.tensordot(theta,np.diag(l[B],0),axes=(3,0))
    
        # Apply U
        theta = np.tensordot(theta,U,axes=([1,2],[0,1]));
 
        # SVD
        theta = np.reshape(np.transpose(theta,(2,0,3,1)),(d*chi,d*chi));
        X, Y, Z = svd(theta); Z = Z.T

        # Truncate
        l[A] = Y[:chi]/np.sqrt(sum(Y[0:chi]**2))
    
        X=np.reshape(X[0:d*chi,0:chi],(d,chi,chi))
        G[A] = np.transpose(np.tensordot(np.diag(l[B]**(-1)),X,axes=(1,1)),(1,0,2));

        Z=np.transpose(np.reshape(Z[0:d*chi,0:chi],(d,chi,chi)),(0,2,1))
        G[B]=np.tensordot(Z,np.diag(l[B]**(-1)),axes=(2,0))
        
    return theta

def representation(G,l,R,chi):
    # Construct the mixed transfermatrix
    
    theta = np.tensordot(G[0],np.diag(l[0],0),axes=(2,0)) # i a b
    theta = np.tensordot(theta,G[1],axes=(2,1)) # i a j b
    theta = np.tensordot(theta,np.diag(l[1],0),axes=(3,0)) # i a j b

    d = R.shape[0]
    O = np.kron(R,R)
    O = np.reshape(O,[d,d,d,d])
    theta_O = np.tensordot(theta,O,axes=([0,2],[0,1])) # a b i j
    
    T = np.tensordot(theta_O,np.conj(theta),axes=([2,3],[0,2])) #a b a' b'
    T = np.transpose(T,(0,2,1,3))
    T = np.reshape(T,(chi**2,chi**2))
    
    # Obtain the dominant eigenvector
    eta,v = eigs(T,which='LM',k=1,ncv=100)
    return eta,np.reshape(v,(chi,chi))
```

```python
######## Define the simulation parameter ######################
chi = 14
d = 3

delta = 0.05
N = 500

D_list = [0,0.25,0.5,0.75,1.25,1.5,2.0]
Sx = np.array([[0,1.,0],[1.,0,1.],[0,1.,0]])/np.sqrt(2)
Sy = np.array([[0,-1j,0],[1j,0,-1j],[0,1j,0]])/np.sqrt(2)
Sz = np.array([[1.,0,0],[0,0,0],[0,0,-1.]])
```

```python
S_list = []
T_list = []
############### Define the initial state: FM state ############
G=[];l=[]

for D in D_list:
    for i in range(2):
        G.append(np.random.rand(3,chi,chi)/10e5); G[-1][1,0,0]=1 
        l.append(np.ones([chi]))       
    
    ########### Define S=1 Heisenberg Bond Hamiltonian and get U ################
    H = np.kron(Sx,Sx)+ np.kron(Sy,Sy) + np.kron(Sz,Sz) + D*np.kron(Sz**2,np.eye(3))
    H = np.real(H)
    
    
    ################# Do the iTEBD run ############################
    
    U = np.reshape(expm(-delta*H),(d,d,d,d))
    for step in range(N):
        theta = update(G,l,U,chi)

    U = np.reshape(expm(-delta/10.*H),(d,d,d,d))
    for step in range(N):
        theta = update(G,l,U,chi)
    
    ################# Measure stuff  ##############################
    S_list.append(-np.inner(np.log(l[0]**2),l[0]**2))

    eta,Vz = representation(G,l,expm(-1j*np.pi*Sz),chi)
    eta,Vx = representation(G,l,expm(-1j*np.pi*Sx),chi)
    T_list.append(np.real(chi*np.trace(np.dot(np.dot(np.dot(Vx,Vz),np.conj(Vx.T)),np.conj(Vz.T)))))
    
pl.figure(figsize=[8,12])
pl.subplot(211)
pl.plot(D_list,S_list,'o-')
pl.ylabel('$S$')
pl.subplot(212)
pl.ylabel('$T$')
pl.plot(D_list,T_list,'o-')
pl.xlabel('$D$')
font = {'size'   : 22}
pl.rc('font', **font)
```
