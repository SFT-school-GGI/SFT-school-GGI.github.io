```python
import numpy as np
import pylab as pl
from scipy import integrate
from scipy.linalg import expm 
```

```python
def update(G,l,U,chi):
    " Updates the G and l matrices using U_bond and the TEBD protocol "
    d = U.shape[0]
    for step in [0,1]:
        A = np.mod(step,2); B = np.mod(step+1,2)

        # Construct theta        
        theta = np.tensordot(np.diag(l[B]),G[A],axes=(1,1))
        theta = np.tensordot(theta,np.diag(l[A],0),axes=(2,0))
        theta = np.tensordot(theta,G[B],axes=(2,1))
        theta = np.tensordot(theta,np.diag(l[B],0),axes=(3,0))
    
        # Apply U
        theta = np.tensordot(theta,U,axes=([1,2],[0,1]));
 
        # SVD
        theta = np.reshape(np.transpose(theta,(2,0,3,1)),(d*chi,d*chi));
        X, Y, Z = np.linalg.svd(theta); Z = Z.T

        # Truncate
        l[A] = Y[:chi]/np.sqrt(sum(Y[0:chi]**2))
    
        X=np.reshape(X[0:d*chi,0:chi],(d,chi,chi))
        G[A] = np.transpose(np.tensordot(np.diag(l[B]**(-1)),X,axes=(1,1)),(1,0,2));

        Z=np.transpose(np.reshape(Z[0:d*chi,0:chi],(d,chi,chi)),(0,2,1))
        G[B]=np.tensordot(Z,np.diag(l[B]**(-1)),axes=(2,0))
        
    return theta

def bond_expectation_value(G,s,O):
    " Expectation value for a bond operator "
    E=[]
    d = G[0].shape[0]
    O = np.reshape(O,(d,d,d,d))
    for step in range(2):
        A = np.mod(step,2); B = np.mod(step+1,2)
        theta = np.tensordot(np.diag(l[B]),G[A],axes=(1,1))
        theta = np.tensordot(theta,np.diag(l[A],0),axes=(2,0))
        theta = np.tensordot(theta,G[B],axes=(2,1))
        theta = np.tensordot(theta,np.diag(l[B],0),axes=(3,0))
        
        theta_O = np.tensordot(theta,O,axes=([1,2],[0,1]))
        E.append(np.squeeze(np.tensordot(np.conj(theta),theta_O,axes=([0,3,1,2],[0,1,2,3]))).item()/np.linalg.norm(theta)) 
    return E
```

```python
######## Define the simulation parameter ######################
chi = 8
d = 2

delta = 0.015
N = 1000
g_list = np.arange(0.1,1.5,0.1)

sx = np.array([[0.,1.],[1.,0.]])
sz = np.array([[1.,0.],[0.,-1.]])
```

```python
S_list = []
m_list = []
dE_list = []

############### Define the initial state: FM state ############
G=[];l=[]
for i in range(2):
    G.append(np.random.rand(2,chi,chi)/10e9); G[-1][0,0,0]=1
    l.append(np.ones([chi]))       
        
for g in g_list:
    ################# Get the exact energies ######################
    f = lambda k,g : -2*np.sqrt(1+g**2-2*g*np.cos(k))/np.pi/2.
    E0_exact = integrate.quad(f, 0, np.pi, args=(g,))[0] 
        
    ########### Define Ising Bond Hamiltonian and get U ################
    H = -np.kron(sz,sz) + g*np.kron(sx,np.eye(2))
    U = np.reshape(expm(-delta*H),(d,d,d,d))
    
    ################# Do the iTEBD run ############################
    for step in range(N):
        theta = update(G,l,U,chi)
    
    ################# Measure stuff  ##############################
    S_list.append(-np.inner(np.log(l[0]**2),l[0]**2))

    E=np.mean(bond_expectation_value(G,l,H))
    dE_list.append(np.abs((E-E0_exact)/E0_exact))
    
    m=np.mean(bond_expectation_value(G,l,np.kron(sz,np.eye(2))))
    m_list.append(m)

pl.figure(figsize=[8,12])
pl.subplot(311)
pl.plot(g_list,m_list,'o-')
pl.ylabel('$m$')
pl.subplot(312)
pl.plot(g_list,S_list,'o-')
pl.ylabel('$S$')
pl.subplot(313)
pl.plot(g_list,dE_list,'o-')
pl.ylabel('$\\Delta E$')
pl.xlabel('$g$')
font = {'size'   : 22}
pl.rc('font', **font)
```
